```python
import pandas as pd

df = pd.read_parquet(
    "../data/raw/raw_data.parquet"
)

print(df.shape)
df.head()
```

```python
df['timestamp_created'] = pd.to_datetime(df['timestamp_created'], unit='s').dt.strftime('%d-%m-%Y')
df['playtime_at_review'] = (df['playtime_at_review'] / 60).round(1)
df.head()
```

```python
df.info()
```

```python
df = df.drop(columns=["Unnamed: 0"])
df = df.rename(columns={'timestamp_created': 'review_date', 'playtime_at_review': 'playtime'})
df.info()
```

```python
df.isna().sum()
```

```python
from langdetect import detect, LangDetectException

def detect_lang(text: str):
    try:
        return detect(str(text))
    except:
        return 'unknown'

df['language'] = df['review'].apply(detect_lang)
```

```python
df.value_counts("language")
```

```python
df = df.drop(df[(df["language"] != "en") & (df["language"] != "so") & (df["language"] != "unknown")].index)
df.value_counts("language")
```

```python
(df['voted_up'].value_counts(normalize=True) * 100).round(1)
```

```python
df.to_parquet("../data/processed/processed_data.parquet")
```
